```python
def rc4(key, data):
    S = list(range(256))
    j = 0

    key = [ord(c) for c in key]

    for i in range(256):
        j = (j + S[i] + key[i % len(key)]) % 256
        S[i], S[j] = S[j], S[i]

    i = 0
    j = 0
    output = []

    for char in data:
        i = (i + 1) % 256
        j = (j + S[i]) % 256

        S[i], S[j] = S[j], S[i]

        k = S[(S[i] + S[j]) % 256]
        output.append(chr(ord(char) ^ k))

    return ''.join(output)


key = "secretkey"
plaintext = "ALTIPARMAK"

encrypted = rc4(key, plaintext)
decrypted = rc4(key, encrypted)

print("Encrypted:", encrypted.encode())
print("Decrypted:", decrypted)
```
